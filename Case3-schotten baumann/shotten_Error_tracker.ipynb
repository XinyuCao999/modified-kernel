```python
import os
import sys
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '..')))
import numpy as np
import plotly.graph_objs as go
from plotly.offline import iplot, plot
from plotly.subplots import make_subplots
import datetime

from BO_base import plot_bo_comparison

# --- Choose which result files to plot (one curve per file) ---
# The number of curves equals the number of files in this list.
# Files live in the BO/ folder. Add / remove files to add / remove curves.
# EI  files:  {model}_{date}_EI.npy
# UCB files:  {model}_{date}_UCB_beta{beta}.npy
date = datetime.date.today().strftime("%m%d")   # e.g. '0726'
date='0726'
beta = 3.5
file_list = [
    f"RBF_zeroprior_{date}_UCB_beta{beta}.npy",
    f"RBF_nonzeroprior_{date}_UCB_beta{beta}.npy",
    f'multi_fidelity_{date}_UCB_beta{beta}.npy',
    f"reconstructed_{date}_UCB_beta{beta}.npy",
    # f"RBF_zeroprior_{date}_EI.npy",
    # f"RBF_nonzeroprior_{date}_EI.npy",
    # f'multi_fidelity_{date}_EI.npy',
    # f"reconstructed_{date}_EI.npy",
]

# Optional custom legend labels (same length as file_list).
# Leave as None to use default names derived from each model type.
label_list = None

# --- Load each file and assemble res_dict_all (one entry per file) ---
res_dict_all = []
info_dict = None
for i, fname in enumerate(file_list):
    d = np.load(os.path.join("BO", fname), allow_pickle=True).item()
    entry = {
        "output_data_all": d["output_data_all"],
        "model_type": d["model_type"],
    }
    if label_list is not None:
        entry["label"] = label_list[i]
    res_dict_all.append(entry)
    if info_dict is None:
        # shared metadata taken from the first file
        info_dict = {
            "res_dict_all": res_dict_all,
            "acquisition_function": d["acquisition_function"],
            "max_iter": d["max_iter"],
            "initial_point": d["initial_point"],
            "beta": d.get("beta", None),
        }
```

```python
plot_bo_comparison(info_dict,reference_max=0.89351549424855,date=date,withdraw_model_index=[],draw_initial_point=True,save_path=None,yaxis_lower_bound=2.5)
```
